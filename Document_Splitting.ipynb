```python
from langchain_community.document_loaders import Docx2txtLoader #reduce space
from langchain_text_splitters.character import CharacterTextSplitter #character splitting
```

```python
loader = Docx2txtLoader("Introduction_to_Data_and_Data_Science.docx")
pages = loader.load()
for i in range(len(pages)):
    pages[i].page_content = ' '.join(pages[i].page_content.split())
```

```python
len(pages[0].page_content)
```

```python
# overlap mean when we split the data previous data end have 2nd phase data 1st line
```

```python
char_spliter = CharacterTextSplitter(separator=".",chunk_size=500,chunk_overlap=50) # we can use chun_overlap=0
```

```python
pages_char_spliter = char_spliter.split_documents(pages)
```
