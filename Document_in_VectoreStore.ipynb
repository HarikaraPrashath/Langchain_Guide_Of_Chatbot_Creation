## Chroma Import Update

In LangChain 1.x, the Chroma integration is maintained in the dedicated package **`langchain-chroma`**.

- Install: `pip install -U langchain-chroma`
- Import: `from langchain_chroma import Chroma`

This avoids the deprecation warning from `langchain_community.vectorstores`.

```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma
from langchain_core.documents import Document
```

```python
embedding = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
#getting data from saved location
vectorstore_from_directory = Chroma(persist_directory = "./intro-to-ds-lectures", 
                                    embedding_function = embedding)
```

```python
vectorstore_from_directory.get()
```

```python
# Retrieve a specific record from the Chroma vector database
# ids = unique ID of the stored document chunk
# include = ["embeddings"] tells Chroma to return the numerical vector
# instead of (or in addition to) the text and metadata
vectorstore_from_directory.get(ids ="62beee79-d3a4-4c5b-a9be-bb42948410fc",  # Unique ID of a stored chunk
                               include = ["embeddings"])                     # Return the stored embedding vector
```

```python
# We manually create a new Document object.
# A Document in LangChain has:
# - page_content → the actual text
# - metadata → extra information describing the text

added_document = Document(
    
    # The main text content that will later be embedded
    # and stored in the vector database
    page_content="Alright! So… Let’s discuss the not-so-obvious differences between the terms analysis and analytics..."
    
    ,
    
    # Metadata provides structured information about this text.
    # This is very important for filtering, search, and context.
    metadata={
        "Course Title": "Introduction to Data and Data Science",
        "Lecture Title": "Analysis vs Analytics"
    }
)
```

```python
# Add a document to the vector store
vectorstore_from_directory.add_documents([added_document])
```

```python
# Then retrieve the document by its ID
vectorstore_from_directory.get("c9bc57d8-b87c-4247-8ab0-7b9984269bbb")
```

```python
#Update the document with new conetent
updated_document = Document(page_content='Great! We hope we gave you a good idea about the level of applicability of the most frequently used programming and software tools in the field of data science. Thank you for watching!', 
                            metadata={'Course Title': 'Introduction to Data and Data Science', 
                                     'Lecture Title': 'Programming Languages & Software Employed in Data Science - All the Tools You Need'})
```

```python
vectorstore_from_directory.update_document(document_id = "c9bc57d8-b87c-4247-8ab0-7b9984269bbb", 
                                           document = updated_document)
```

```python
vectorstore_from_directory.get("c9bc57d8-b87c-4247-8ab0-7b9984269bbb")
```

```python
vectorstore_from_directory.delete("c9bc57d8-b87c-4247-8ab0-7b9984269bbb")
```

```python
#now nothing to show becuase that was delted 
vectorstore_from_directory.get("c9bc57d8-b87c-4247-8ab0-7b9984269bbb")
```
